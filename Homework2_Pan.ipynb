```python
import random

#List of random integers
def generate_random_list(size=50, min_val=1, max_val=100):
    return [random.randint(min_val, max_val) for _ in range(size)]

#List in ascending order
def bubble_sort(arr):
    n = len(arr)
    for i in range(n - 1):
        for j in range(n - 1 - i):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]

#Looks at the element to the right of each position
def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        while j >= 0 and key < arr[j]:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key

#Lists
random_list = generate_random_list()
print("50 Element list:")
print(random_list)

bubble_sorted_list = random_list.copy()
bubble_sort(bubble_sorted_list)
print("Bubble Sort:")
print(bubble_sorted_list)

insertion_sorted_list = random_list.copy()
insertion_sort(insertion_sorted_list)
print("Insert Sort:")
print(insertion_sorted_list)
```
